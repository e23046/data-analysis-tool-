## **Step 2: Generate Dummy Data & Initialize Toolkit**
Let's simulate a messy engineering or business dataset containing missing entries (`NaN`), extreme outlier variations, and unhandled categorical labels.

```python
import numpy as np
import pandas as pd
from data_analysis import DataInspector, PlottingMethods

# Create a simulated messy dataset
np.random.seed(42)
n_samples = 150

data = {
    "Age": np.random.choice([25, 30, 45, 50, np.nan, 200], size=n_samples), # Outlier (200) and NaNs
    "Salary": np.random.choice([45000, 60000, 85000, np.nan, 120000, 1000000], size=n_samples), # Outlier (1000000)
    "Department": np.random.choice(["Engineering", "HR", "Marketing", "Sales", None], size=n_samples),
    "Performance_Score": np.random.uniform(1.0, 5.0, size=n_samples)
}

df_raw = pd.DataFrame(data)

# Initialize our custom package classes
inspector = DataInspector()
pm = PlottingMethods()

# Explicitly load the generated DataFrame into the inspector toolkit
inspector.df = df_raw.copy()
print("✅ Inspector initialized with structural dummy data!")
inspector.df.head()
```

## **Step 3: Data Inspection & Structural Summaries**
Examine missing row instances, data type profiles, column boundaries, and categorical spreads.

```python
print("=== Dataset Structural Summary ===")
inspector.get_summary()

print("\n=== Column Type Details ===")
inspector.column_details()

print("\n=== Categorical Missing Masks ===")
inspector.get_categorical_summary()
```

## **Step 4: Automated Data Cleaning & Outlier Management**
Impute unassigned values via strategic feature medians, clear structural row duplicates, and eliminate outliers using the Interquartile Range ($IQR$) threshold boundaries.

```python
# Initialize and prompt for ANY file upload (Titanic, Employee, Sales, etc.)
inspector = DataInspector()
inspector.upload_forced_file()

# 1. Automatically handle missing numbers on all numeric columns
inspector.handle_missing_values(strategy='median')

# 2. Automatically handle remaining categorical null blanks on all text columns
inspector.handle_missing_values(strategy='mode')

# 3. Clean duplicates safely
inspector.remove_duplicates()

# 4. Automatically identify and remove outliers from all numeric columns present
inspector.handle_outliers()

print("\nUpdated Dataset Final Dimensions:", inspector.df.shape)
```

## **Step 5: Preprocessing, Scaling, & Feature Encoding Pipelines**
Extract continuous features for normalization algorithms (`MinMaxScaler`, `StandardScaler`) and convert categorical descriptors into multi-column hot vectors.

```python
print("--- Extracted Normalized Numeric Array Preview ---")
numeric_scaled = inspector.extract_normalized_numeric_data(method='minmax')
print(numeric_scaled.head())

print("\n--- Extracted One-Hot Categorical Features Preview ---")
categorical_encoded = inspector.extract_normalized_categorical_data(method='onehot')
print(categorical_encoded.head())

print("\n--- Unified Combined Engine DataFrame Summary ---")
ml_ready_df = inspector.create_normalized_data_df()
print(f"Total pipeline features mapped: {ml_ready_df.shape[1]}")
```

## **Step 6: Statistical Matrix Verifications & Associations**
Run stationarity analysis, variance stability metrics, sequence autocorrelation validations, and unified heatmaps across features.

```python
# 1. Test Multivariate Mean Stability via Wilks' Lambda MANOVA approximations
inspector.test_constant_mean(chunks=5)

# 2. Test Covariance Matrix Stability via Box's M tests
inspector.test_constant_covariance(chunks=5)

# 3. Check Sequence Dependency via Multivariate Ljung-Box autocorrelation metrics
inspector.test_row_independence(max_lag=4)

# 4. Render the Unified Multi-Type Feature Association Strength Map
print("\n=== Processing Combined Feature Relationship Map ===")
inspector.plot_all_associations_heatmap()
```
